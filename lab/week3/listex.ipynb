## LIST 연습 노트 파일입니다

리스트의 합병 연산 (+) 리스트의 곱 연산 (*)

```python
fruit1 = ["사과", "바나나"]
fruit2 = ["복숭아", "포도"]
print(fruit1)
```

```python
fruits = ["사과", "바나나", "체리"]
fruits.remove("바나나")
print(fruits)
```

```python
fruits = ["사과", "복숭아", "체리"]
fruits.insert(1, "딸기")
print(fruits)
```

```python
fruits.append("샤인머스캣")
print(fruits)
```
