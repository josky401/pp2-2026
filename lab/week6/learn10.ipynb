# 작업폴더 python 파일 실행하는 위치

```python
with open("text.txt", "w") as f:
    f.write("Hello, World!")
```

```python
with open("text.txt", "r", encoding="utf-8") as f:
    #for line in f:
    #    print(line.strip())

    lines = f.readline().strip() 
    while lines!= "":
        print(lines)
        lines = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    f.seek(0, 0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
```

# 문법

# 예외처리

try: 검사할 코드 except(오류명): 오류대신 반환내용 else: 오류가 없다면 주로 탈출(break)사용 finally: 오류가 발생했든 안했든 실행
