# 내장함수 실습

- 절댓값 abs(대상)

- and 연산, bool출력 all(리스트 또는 딕셔너리)

- or 연산, bool출력 any(위와 동일)

- 수식계산 eval(문자열 수식)

- map(리스트나 튜플) (변수 <- 리스트) 리스트에서 하나씩 변수를 거친 결과물 기본출력id 리스트 씌워야함

- dir(타입예시) 변수,함수출력

- enumerate() 인덱스 0부터 튜플로 붙여줌 ...ex enumerate(listname, start=1)

- filter(변수, 리스트튜플) 출력은 id덩어리로 나오니 리스트로 출력하도록 작성필요

- zip(리스트1, 리스트2) 튜플로 묶여나옴 리스트변환필요 반복문 가능 -> for l1, l2 in zip(리스트1, 리스트2): print(l1, l2)
